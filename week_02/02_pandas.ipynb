```python
# Create a simple Pandas Series from a list:

import pandas as pd

a = [1, 7, 2]

b = pd.Series(a)

print(b)
print("--------")
print(b[0])
```

```python
# we can create the labels for the series by passing a list of labels to the index parameter:
a = [1,2,3,4,5,6]

b = pd.Series(a, index=['a','b','c','d','e','f'])

print(b)
print("--------")
print(b['f'])
```

```python
# create a pandas Series from a dictionary:
a = {'a': 90, 'b': 120, 'c': 200, 'd': 250}

b = pd.Series(a)
print(b)
```

```python
# create a simple Pandas Series from a dictionary:

import pandas as pd

calories = {"day1": 420, "day2": 380, "day3": 390}

myvar = pd.Series(calories, index= ['day1', 'day2'])

print(myvar)
```

# DataFrame:
A `DataFrame` is a two-dimensional labeled data structure with columns of potentially different types.

Data sets in Pandas are usually multi-dimensional tables, called `DataFrames`.

`Series` is like a column, a DataFrame is the whole table.

```python
import pandas as pd
data = {
    'days': [1,2,3,4,5,6,7],
    'month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul']
}

df = pd.DataFrame(data)
df.head()
```

```python
#refer to the row index:
print(df.loc[0])
```

```python
# use the list of indexes to return the multiple rows:
print(df.loc[[0,1,2]])
```

```python
# Add a list a names to give each row a name:
df = pd.DataFrame(data, index = ['a', 'b', 'c', 'd', 'e', 'f', 'g'])

df
```

```python
# used the name of the index to refer to the row
print(df.loc['a'])

print('-----')
# for multiple rows, use the list of names

print(df.loc[['a', 'b', 'c']])
```
