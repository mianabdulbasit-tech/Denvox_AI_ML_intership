```python
# Program to check if a number is prime and if a string is a palindrome

# Check prime number
num = int(input("Enter a number: "))

if num <= 1:
    print("Not Prime")
else:
    for i in range(2,  num):
        if num % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")

# Check palindrome
str = input("Enter a string: ")
str = str.lower()

reversed_str = str[::-1]

if str == reversed_str:
    print("Palindrome")
else:
    print("Not palindrome:")
```

```python
# task 3

# Print the Fibonnacci series using loop

def fib(num):
    a , b = 0 , 1
    while a < num:
        print(a, end=' ')
        a, b = b , a+b
    print()

fib(5)
```
